6. Vad är ”klassificieringsproblem? Kan du ge några exempel på modeller som används och potentiella tillämpningsområden?

Tillämpningsområden kan ligga inom kundanalys där man t ex kan predikera om en kund kommer att handla igen eller inte (ett annat ord för detta är churna). Detta baseras på data som antal köp, ålder, kön och antal supportärenden.

7. Vad är en ”Confusion Matrix”?
Svar: Confusion Matrix är ett verktyg vid just klassificeringsproblem för att se hur bra en modell presterar. Detta görs genom att visuellt jämföra sanna och predikerade värden. Det är en form av matris med 4 kategorier (för binära klassificeringsprobelm). True positive (TP) = Antal gånger modellen prediker en positiv klass korrekt, False Negative (FN) = Antalet gånger en modell felaktigt predikerar en negativ klass, False Positive (FP) = Antalet gånger en modell felaktigt predikerar en positiv klass och True Negative (TP) = Antalet gånger en modell korrekt predikerar en negativ klass.
Vid multiklassprobelm så blir denna confusion matrix mycket större. I den matrisen har man alltså samma klass både vertikalt och horizontelt med sanna resp. predikerade värden. Då kan man se antalet gånger som modellen korrekt predikerade korrekt resp. hur många gånger den predikerade felaktigt en annan klass.

8. Hur definieras Precision? Hur tolkas Precision?
Svar: Precision är ett utvärderingsmått som mäter andelen av de positiva prediktionerna som faktiskt är korrekta. Man beräknar precision genom formeln TP/(TP + FP) = Precision
Om en modell har ett högt värde för Precision så innebär detta t ex på tillämpningen med t ex sjuka patienter att vi predikerar många som egentligen är sjuka som friska. Vid allvarlig sjukdom är inte detta önskvärt. Vad man egentligen vill ha är ett lägre värde på precision men ett högre värde på Recall där man faktisk predikerar fler som sjuka även om de visar sig vara friska. På det sättet missar vi färre svårt sjuka patienter. 

9. Hur definieras Recall? Hur tolkas Recall?
Svar: Recall hänger ihop med Precision som vi talade om ovan. Man beräknar Recall genom formeln TP/(TP + FN) = Recall. Vad man kan se här är att det finns en trade-off mellan Recall och Precision. Antingen är vi mer träffsäkra med att Precision ökar eller så breddar vi scopet och ökar Recall där vi då får med flera falska positiva. Men, vid fall där det är mer viktigt att inte missa positiva fall är ett högre Recall att föredra.

11. Vad är skillnaden mellan parametrar och hyperparametrar i en maskininlärningsmodell? Ge ett exempel på varje och förklara varför de inte kan optimeras på samma sätt.
Svar: Kort definition av parametrar är "vad modellen lär sig under träning". Hyperparametrar är mer "Hur modellen lär sig". Hyperparametrar påverkar hela träningsprocessen och man kan testa olika varianter med t ex GridSearch.
Ett konkret exempel kan vara vid linjär regression (Y = B0 + B1X) där parametrarna B0 och B1 styr intercept och lutning. Under träning skattas dessa båda värden. En hyperparameter är mer en inställning Hur en modell ska tränas.

12. Förklara skillnaden mellan overfitting och underfitting i en maskininlärningsmodell. Beskriv även hur man kan upptäcka respektive åtgärda dem.
Svar: 
Underfitting innebär att modellen är för enkel för att fånga de verkliga mönstren i datan.
Exempelvis kan man tänka an vanlig linjär regressionsmodell med en diagonal linje där datan ser helt annorlunda ut. Datan är t ex mer kvadratisk och då behöver man en polynomregression istället.

Overfitting innebär då att modellen är för komplex för datan och fungerar bra på träningsdata med fungerar sämre på verklig osedd data. Modellen uppfattar brus som mönster och anpassar sig alltså till detta mönster som den egentligen ska ignorera. Man kan säga att modellen memorerar istället för att generalisera.

Overfitting upptäcks när modellen har lågt träningsfel men högt testfel, vilket visar att den inte generaliserar. Underfitting upptäcks när både tränings- och testfel är höga, vilket tyder på att modellen är för enkel. Detta analyseras genom att jämföra prestanda på tränings- och testdata, ofta med hjälp av train-test split eller korsvalidering.

För att åtgärda overfitting brukar man titta på att göra modellen enklare genom färre parametrar, Regularisering (L1/Lasso, L2/Ridge) som straffar stora koefficiener, mer data för att minska risken att modellen memorerar, Cross validation för att hitta rätt hyperparametrar. Ett annat exempel är early stopping stoppa innan modellen börjar memorera.

För att åtgärda underfitting brukar man titta på att öka modellens komplexitet med fler features och djupare träd. Man kan träna modellen längre (fler iterationer) sist tittar man även på bättre features sk. feature engineering.

13. Hur funkar OvO och OvR algoritmerna?
Svar: OvO betyder One-vs-One och OvR betyder One-vs-Rest och används av binära klassificeringsmodeller för att med dessa algoritmer kunna utvidgas till att även kunna hantera multiklass klassificering.

För OvR kan man anta att vi ska klassificera siffrorna 0-9. Då har man en klass för varje siffra där varje modell predikerar om siffran är 0, 1, 2, 3 osv. 

För OvO tränar man istället en modell för varje par av klasser. Det innebär att om det finns N antal klasser kommer (N X (N-1))/2 stycken modeller att behöva tränas. Varje modell jämför ett klasspar och predikerar en av de två klasserna. Den klass som predikeras flest gånger väljs slutligen.

14. Vad är intuitionen bakom att regularisera en modell?
Svar: Regulaisering handlar om att motverka överanpassning genom att minska värdet på vikterna. Genom att man minskar värdet på vikterna så begränsar man modellen och gör den mindre flexibel. Detta ökar bias men minskar variansen. Här är förhoppningen att variansen blir större än vad ökningen av bias blir.

Intutionen bakom regularisering görs med ett exempel där man har en budget på hur stora vikterna för våra oberoende variabler får vara. I det fallet innebär det att de variabler som inte är tillräckligt användbara sätts till 0. Detta för att spara budgeten till de variabler som i högre utsträckning bidrar till att vi kan genomföra bra prediktioner.

15. Vad är Bias Variance Trade-off?
Svar: Bias Variance Trade-off kan delas in i 3 delar:
- Bias: Det är när vi antar felaktiga modellantaganden. Exempelvis kan vi anta att datan är linjär medans den egentligen är kvadratisk. Detta kan då alltså leda till underanpassning.
- Variance: Detta fel beror på att modellen är känslig för fluktuationer i träningsdatan när den tränas. Detta i sig kan leda till att överanpassning
- Irreducerbart fel: Det här är ett fel som beror på att vi har slumpmässighet i datan. Detta är inget vi kan eller ska göra något åt, det är så datan ser ut. 

Generellt sett, när vi går från en enkel modell till en mer komplex modell så ökar dess varians och minskar bias och vice versa. Det är därför inte alltid bättre med en mer komplex modell eftersom ökningen i varians kan kosta mer än det smakar i minskning av bias. Detta är anledningen till att man kallar det Bias Variance Trade-off.

16. Varför är mer komplexa modeller inte alltid bättre?
Svar: En komplex modell kan börja memorera träningsdatan istället för att lära sig generella mönster. En allt för komplex modell lär sig "spam". I dessa fall fungerar modellen bra på träningsdata med dåligt på ny osedd data. Detta är ett central problem inom maskininlärning.
Stora modeller har enormt många parametrar, vi kan jämföra med en linjär regression som kräver några få parametrar medans ett neuralt nätverk kan ha många tusentals eller även miljarder parametrar. 
I det fallet med neurala nätverk och datamängden inte är tillräckligt stor så får den svårt att lära sig robusta samband. 

En liten modell med bra data kan därför slå en större modell med otillräcklig data.
