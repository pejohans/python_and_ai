15. Vad är Bias Variance Trade-off?
Svar: Bias Variance Trade-off kan delas in i 3 delar:
1. Bias: Det är när vi antar felaktiga modellantaganden. Exempelvis kan vi anta att datan är linjär medans den egentligen är kvadratisk. Detta kan då alltså leda till underanpassning.
2. Variance: Detta fel beror på att modellen är känslig för fluktuationer i träningsdatan när den tränas. Detta i sig kan leda till att överanpassning
3. Irreducerbart fel: Det här är ett fel som beror på att vi har slumpmässighet i datan. Detta är inget vi kan eller ska göra något åt, det är så datan ser ut. 

Generellt sett, när vi går från en enkel modell till en mer komplex modell så ökar dess varians och minskar bias och vice versa. Det är därför inte alltid bättre med en mer komplex modell eftersom ökningen i varians kan kosta mer än det smakar i minskning av bias. Detta är anledningen till att man kallar det Bias Variance Trade-off.
