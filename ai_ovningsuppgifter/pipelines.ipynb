```python
import numpy as np
from sklearn.impute import SimpleImputer
```

```python
A = np.array([[np.nan, np.nan, 2], [4, np.nan, 1], [10, 5, 9]])
print(A)
```

```python
#Compute the mean of each column, ignoring NaN values
np.nanmean(A, axis=0)
```

```python
imp_mean = SimpleImputer(missing_values=np.nan, strategy='mean')

A_transformed = imp_mean.fit_transform(A)
A_transformed
```

```python
from sklearn.preprocessing import PolynomialFeatures
```

```python
# Create a polynomial feature transformer with degree 2
poly = PolynomialFeatures(2)
poly.fit_transform(A_transformed)



```

```python
#Doing the steps above in a pipeline
from sklearn.pipeline import Pipeline
pipeline = Pipeline([
    ('imputer', SimpleImputer(missing_values=np.nan, strategy='mean')),
    ('poly', PolynomialFeatures(2))
])

C = pipeline.fit_transform(A)
C
```
