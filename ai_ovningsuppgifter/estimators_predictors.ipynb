```python
import numpy as np
import matplotlib.pyplot as plt 
import pandas as pd

```

```python
#Old ways of generating data, see next cell for better way to do it.
np.random.seed(42)
X = 2 * np.random.rand(100, 1)
y = 4 + 3 * X + np.random.randn(100, 1)
print(y)


#print(X)
```

```python
#New way of generating data, using the new random generator.
rng = np.random.default_rng(42)
#print(type(rng))
X = 2 * rng.random((100, 1))
#print(X)
y = 4 + 3 * X + rng.normal(size=(100, 1))
#print(y)

plt.scatter(X, y)



```

```python
from sklearn.linear_model import LinearRegression

#Create a linear regression model and fit it to the data
lin_reg = LinearRegression()
lin_reg.fit(X, y)
```

```python
lin_reg.get_params()
```

```python
print(lin_reg.coef_, lin_reg.intercept_)
```

```python
X_new = np.linspace(0, 2, 20).reshape(-1, 1)
#print(X_new)
```

```python
#Use lin_reg as a predictor to predict the values of y for the new X values.
y_pred_lr = lin_reg.predict(X_new)

```

```python
fig, ax = plt.subplots()
ax.set_title('Linear Regression')
ax.scatter(X, y, label='Data')
ax.plot(X_new, y_pred_lr, color='red', label='Linear Regression', linewidth=2, linestyle='--')
ax.legend()
```
