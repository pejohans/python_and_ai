3. Förklara följande:
a) Vad är syftet med att dela upp data i träningsdata, valideringsdata och
testdata?
Svar: Syftet är att man vill träna och utvärdera olika modeller på osedd data. Efter att man provat lite olika modeller på träningsdata så väljer man ut de som presterar bäst och validerar dessa på valideringsdata. Om modellerna presterar bra även på denna osedda data så går man vidare med den som presterat bäst och tester på osedd testdata. Innan man gör det så ska man dock träna om modellen på både träningsdata och valideringsdata. I slutändan efter att man sett att modellen presterar enligt önskemål på osedd testdata så går man vidare och tränar modellen på hela datasettet innan man går i produktion.

b) Vad är k-delad korsvalidering för något?
Svar: K-delad korsvalidering är ett annat sätt att träna en modell på. Man tränar och validerar samtidigt i iterationer på olika delar av datasettet. Antag att vi sätter CV=5 så innebär detta att vi delar upp datasettet i 5 delar och under 5 iterationer så tränar vi och validerar vi på hela datasettet. Detta är vad jag förstår ett populärt val eftersom det ger bra resultat. Nackdelen kan vara att det kan ta lång tid och kanske inte alltid lämpligt.

c) Vad är root mean squared error (RMSE) för något?
Svar: RMSE är ett värde på andelen fel som en modell producerar. Detta värde används i syfte om utvärdera vilken modell som presterar bäst. 

d) Vad är en hyperparameter för något? Vad är en parameter för något?
Svar: Hyperparametrar är värden som styr hur modellen lär sig och som inte lärs av modellen själv under träning. 

e) Vad är grid search och hur hänger namnet “grid” och “search” ihop med
processen som genomförs? Läser vi dokumentationen: https://scikit-learn.
org/stable/modules/generated/sklearn.model_selection.GridSearchCV så
ser vi att hyperparametern refit har standardvärdet True, vad innebär
det?
Svar: Grid search är ett automatiserat sätt att välja hyperparametrar på. Det som algoritmen gör är att den för varje hyperparameter tillämpar k-delad korsvalidering och på så vis ser vilken kombination som är optimal. 

"grid" och "search" hänger ihop på så sätt att "grid" utgörs av en lista av hyperparametrar och dess möjliga värden. "Search" innebär att man gör en sökning över alla kombinationer för att att hitta den bästa. 

Att refit=True innebär att efter att vi genomfört vår grid search och kört .fit() så kommer modellen att tränas om på de bästa hyperparametrarna.

f) Vad är kategorisk data och hur hanteras det? I ditt svar, använd begreppen
nominal data, ordinal data, one-hot-encoding, dummy-variable-encoding
och ordinal encoding.
Svar: Katogorisk data är data som är indelad i olika kategorier. T ex kön, som man eller kvinna. När vi arbetar med ML-modeller så behöver dessa numerisk data och här börjar man då tala om nominal data och ordinal data. Dessa båda två har destinkta skillnader genom att nominal data är att fixt antal kategorier utan inbördes rangordning (men det beror på hur det tolkas) medan ordnial data är rangordnad. Vidare talar man om att göra om nominal data till ordinal data och då kommer vi in på ordinal encoding som är ett sätt att omvandla kategorisk data till numerisk data. 

Så, om vi nu har nominal data som vi vill göra om till ordinal data, t ex färgerna {röd, grön, blå, svart} så behöver vi genomföra en sk. one-hot-enconding genom att använda oss av funktionen get_dummies(nominal_df, dtype=int). Vad vi då får är ett dataset som innehåller en kolumn för varje färg och där varje rad innehåller 0 eller 1 för kolumnerna. T ex för kolumnen röd så sätts en 1:a för att indikera på färge röd, alla de andra kolumnerna får värdet 0.

Sista uttrycket är dummy-variable-encoding och innebär att för varje k=kategori får vi k-1 kolumner. Det vill säga att i fallet med färger så får vi kolumnerna röd, grön och blå. Detta innebär att om vi sätter en 1:a för röd, grön och blå kolumn så indikerar detta på färgen svart. Med andra ord, vi utnyttjar summan av flera kolumner.

Anledningen till att man använder sig av dessa metoder är att man vid den linjära regressions-modellen använder sig av dummy-variable-encoding och i andra fall använder man sig av one-hot-encoding.

g) Vad är feature engineering för något?
Svar: Feature engineering innebär i princip att du som modellerare väljer ut relevanta features/värdemängder som är relevanta att träna modeller på. För mycket variabler riskerar att göra modellen övertränad och kommer att prestera dåligt på osedd data. 
Detta är något av en konst och man behöver utveckla en känsla för vilka värdemängder/kolumner som man ska ta med vid träning av en modell. Det är även viktigt att känna till själva domänen. Dvs. Antag att du ska förutsäga priser på begagnade bilar och till vilka målgrupper som du ska sälja vissa typer av bilar till så behöver du vara påläst kring marknaden för begagnade bilar. 

h) Vad menas med principle of parsimony
Svar: Principle of parsimony är en generell princip inom modellering. Den kallas även Ockhams rakkniv. Vad denna generella princip handlar om är att givet en viss prestation så försöker man ha en så enkel modell som möjligt.
