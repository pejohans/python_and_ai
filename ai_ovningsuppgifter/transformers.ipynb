```python
import numpy as np
import matplotlib.pyplot as plt 
import pandas as pd
```

```python
#New way of generating data, using the new random generator.
rng = np.random.default_rng(42)
X = 2 * rng.random((100, 1))
y = 4 + 3 * X + rng.normal(size=(100, 1))

plt.scatter(X, y)
```

```python
print(np.mean(X), np.var(X))
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaler.fit(X)
```

```python
print(scaler.mean_, scaler.var_)
```

```python
X_transformed = scaler.transform(X)
```

```python
print("{:.2f}".format(np.mean(X_transformed)))
print(np.var(X_transformed))
```

```python
scaler_2 = StandardScaler()
X_transformed_2 = scaler_2.fit_transform(X)
```

```python
print("{:.2f}".format(np.mean(X_transformed_2)))
print(np.var(X_transformed_2))
```
