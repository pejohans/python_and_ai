Svar: Generellt så arbetar man alltid iterativt fram och tillbaka mellan stegen, precis som i de flesta utvecklingsprojekt. De 7 stegen är följande:
- Definiera problemet och skapa en helhetsbild - Vad är målet med projektet, sätt rätt förväntningar, vilket värde tillförs till organisationen?, vad finns för lösningar inom organisationen idag som kan ge oss liknande resultat?

- Få tillgång till datan - Hur får vi tillgång till datan och vilken data finns? Är datan av könslig karraktär och behöver vi då anonymisera data? Om vi upptäcker att vi t ex inte har tillräckligt med data för att uppnå målet med projektet, tillbaka till steg 1 och omdefiniera målet med projektet. Lägg undan testdata redan nu.

- Utforska datan (EDA) - Gör en EDA på kopia av datan för att se vad som går att göra med den data vi har. Kanske fler upptäckter görs som ställer nya frågor och som kanske rent av är relevanta för organisationen. Vidare ger oss EDA möjlighet att se mönster i datan, genomföra visualiseringar, dra slutsatser om datan och vilka kopplingar som finns i datan. Vi får även information om vilken data som kanske saknas och om vi behöver rensa datan pga irrelavant data. 

- Bearbeta datan - Även i detta steg, som också kallas "Data cleaning" arbetar vi med en kopia av datan. I detta steg skapar vi funktioner för tranformering av datan som vi senare kan använda igen och igen. Kan vara så att ny uppdaterad data kommer från kunden och då kan vi lättare transformera ny data. 
Transformering av datan innebär att vi kanske rensar bort null-värden eller kanske ersätter vissa null-värden med medelvärden osv. Det kan även vara så att vi i detta steg hittar samband som gör att vi skapar helt nya kolumner(feature engineering) med hjälp av andra kolumners data. 

- ML-modellering - I detta steg utvärderar vi olika modeller på träningsdatan. Därefter validerar vi den modell vi valt på valideringsdatan. Till slut testar vi modellen på osedd testdata. 

- Presentera din lösning för intressenter - I detta steg presenterar vi lösningen för olika intressenter. Här behöver vi fundera på hur vi formar vår presentation och vilken abstraktionsnivå vi lägger oss på. För utvecklare/teamet så är det bra att lägga sig på en mer teknisk nivå eftersom de med största sannorlikhet vill veta detaljer i lösningen. I de fall då du presenterar för chefer och organisationen som inte är så tekniskt lagda så behöver du presentera på en mer övergripande nivå, gärna med mer grafer för att visualisera resultatet. En bild säger mer än 1000 ord. 

- Produktionssättning och övervakning av modellen och implementeringen - I detta sista steg (enligt läroboken - Jag personligen känner att man behöver ytterligare ett steg innan detta steg då man gör en release till testmiljö där man faktiskt kan be kunden/referensgrupp testa lösningen under en tid innan man går i produktion. Detta ger också en flexibilitet att testa nya tillämpningar innan man går i produktion) så tar man fram enhetstester och en strategi för att övervaka modellen. Modeller tenderar till att prestera sämre med tiden, oftast beroende på att data förändras. Då behöver man träna om modellen på ny data. Det är inte ovanligt att man tränar om modellen veckovis eller månadsvis.

```python
from sklearn.linear_model import LinearRegression
from sklearn.datasets import make_regression
from sklearn.model_selection import train_test_split
import joblib
import matplotlib.pyplot as plt
from sklearn.metrics import r2_score

x, y = make_regression(n_samples=100, n_features=1, noise=10, random_state=42)
print(f'X Shape before split {x.shape}')
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
print(f'X Shape after split {x_train.shape}')

model = LinearRegression()
model.fit(x_train, y_train)

#prediction
y_pred = model.predict(x_test)
print(f'y_pred={y_pred}')

#Score
print(f'R2 Score:{r2_score(y_test, y_pred)}')

print(f'model.coef_ = {model.coef_}')
print(f'model.intercept_ = {model.intercept_}')

#plot data + regressionline
plt.scatter(x_test, y_test)
plt.plot(x_test, y_pred)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Linear Regression fit on training data")

plt.show()

#plot true vs predicted
plt.scatter(y_test, y_pred)
plt.xlabel("True values")
plt.ylabel("Predicted values")
plt.title("True vs predicted")

plt.show()

#Residual plot
residuals = y_test - y_pred
print(f'residuals={residuals}')
plt.scatter(y_pred, residuals)
plt.axhline (y=0)
plt.xlabel("Predicted values")
plt.ylabel("Residuals")
plt.title("Residuals plot")

plt.show()

joblib.dump(model, 'our_linear_model.pk1')
print('Model saved...')

loaded_model = joblib.load('our_linear_model.pk1')
print('Model loaded...')

prediction = loaded_model.predict([x_test[5]])
print(f'Feature value: {x_test[5]}. Predicted valude: {prediction}')
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.metrics import root_mean_squared_error
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import GridSearchCV

housing_original = pd.read_csv('housing.csv')
housing_original.info()

housing_original.iloc[1:5, 1:3]

print(housing_original['ocean_proximity'].value_counts())

housing = housing_original[housing_original['ocean_proximity'] != 'ISLAND']
print(housing['ocean_proximity'].value_counts())

housing = pd.get_dummies(housing, columns=['ocean_proximity'], dtype=int, prefix='dmy')
housing.iloc[[1,200,1000,1800,3000,3500], 9:]

train_full, test = train_test_split(housing, test_size=0.2, random_state=40)
train, val = train_test_split(train_full, test_size=0.25, random_state=36)

print('Ta bort rader som har kolumner med saknade värden')
train = train.dropna()
print('Träningsdata...')
train.info()

#Droppa även rader med tomma värden i validering och test-dataseten
val = val.dropna()
test = test.dropna()




```

```python
X_train_full = train_full.drop(columns=['median_house_value'])
y_train_full = train_full['median_house_value']

print(train_full.shape)

X_train, y_train = train.drop(columns=['median_house_value']), train['median_house_value']
X_val, y_val = val.drop(columns=['median_house_value']), val['median_house_value']
X_test, y_test = test.drop(columns=['median_house_value']), test['median_house_value']

print(X_train.shape)
print(y_train.shape)
```

```python
from sklearn.linear_model import LinearRegression

lin_reg = LinearRegression()
lin_reg.fit(X_train, y_train)

import time
from sklearn.ensemble import RandomForestRegressor

start_time = time.time()

rf = RandomForestRegressor()
hyperparam_grid = {'max_depth': [5,10,15,50], 'n_estimators': [1,5,10]}
grid_search = GridSearchCV(rf, hyperparam_grid, scoring='neg_root_mean_squared_error', cv=5)
grid_search.fit(X_train, y_train)

end_time = time.time()

execution_time = end_time - start_time
print(f'GridSearchCV fitting took {execution_time:.4f} seconds')
```

```python
print('Best Hyperparameters from GridSearchCV:', grid_search.best_params_)
pd.DataFrame(grid_search.cv_results_)
```

```python
lr_pred_val = lin_reg.predict(X_val)
rf_pred_val = grid_search.predict(X_val)

print('RMSE Linear Regression:', root_mean_squared_error(y_val, lr_pred_val))
print('RMSE Random forest Regression:', root_mean_squared_error(y_val, rf_pred_val))
```
